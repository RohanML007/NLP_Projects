```python
from transformers import pipeline, AutoTokenizer, AutoModelForSeq2SeqLM

#"pipeline" is a high-level API for performing NLP tasks with a single line of code.
#You don’t need to manually load models or tokenizers—it does everything for you.
#Supports tasks like translation, text summarization, sentiment analysis, etc..

#"AutoTokenizer" loads the correct tokenizer for any given transformer model.
#A tokenizer converts text into numerical tokens before feeding it to a model.
#Automatically selects the correct tokenization technique based on the model.

#"AutoModelForSeq2SeqLM" loads a pre-trained sequence-to-sequence model (used for translation, summarization, etc.).
#Unlike pipeline(), this gives full control over model usage.
#You need to manually tokenize input, pass it to the model, and decode the output.


import pickle

import pandas as pd

import csv

from datasets import load_dataset

import matplotlib.pyplot as plt

import seaborn as sns
```

### **Downloading the "Helsinki-NLP/opus-mt-en-hi" machine translation model:**

```python
translator_opus = pipeline("translation_en_to_hi", model="Helsinki-NLP/opus-mt-en-hi", device=-1)
#"translation_en_to_hi" is a task name to be handled by the model
#"device=-1" means we are allowing it to use the CPU and if "device=0", it means it can use GPU for the processing.

english_sentence = input("Enter an english sentence : ")

hindi_translator_opus = translator_opus(english_sentence, max_length=50)[0]['translation_text']

print("Translation (opus):", hindi_translator_opus)
```

### **Downloading the "facebook/m2m100_418M" machine translation model:**

```python
from transformers import pipeline, AutoTokenizer, AutoModelForSeq2SeqLM

translator_m2m = pipeline("translation_en_to_hi", model="facebook/m2m100_418M", device=-1)

english_sent = input("Enter the english sentence here : ")

hindi_translator_m2m = translator_m2m(english_sent, max_length=50)[0]['translation_text']

print(f"Translation m2m : {hindi_translator_m2m}")
```

### **Downloading the "facebook/mbart-large-50-many-to-many-m-model" machine translation model:**

```python
from transformers import pipeline, AutoTokenizer, AutoModelForSeq2SeqLM

translator_mbart = pipeline("translation", model="facebook/mbart-large-50-many-to-many-mmt", device=-1, src_lang="en_X",
                            tgt_lang="hi_IN")

english_sentence1 = input("Enter the english sentence here : ")

hindi_mbart = translator_mbart(english_sentence1, max_length=50)[0]['translation_text']

print(f"Translation (mbart): {hindi_mbart}")
```

### **Calculating the BLEU Score for each of the above models:**

```python
merged_df_subset.columns
```

```python
from sacrebleu import corpus_bleu

original_translations = merged_df_subset['Hindi_Sentence'].tolist()

opus_translations = merged_df_subset['Hindi_opus'].tolist()
m2m_translations = merged_df_subset['Hindi_m2m'].tolist()
mbart_translations = merged_df_subset['Hindi_mbart'].tolist()

#Converting the model translations to string format
opus_translations = [str(x) for x in opus_translations]
m2m_translations = [str(x) for x in m2m_translations]
mbart_translations = [str(x) for x in mbart_translations]

#Lets calculate the BLEU Score now
bleu_opus = corpus_bleu(opus_translations, [original_translations])
bleu_m2m = corpus_bleu(m2m_translations, [original_translations])
bleu_mbart = corpus_bleu(mbart_translations, [original_translations])

print(f"BLEU Score for Opus-MT: {bleu_opus.score}")
print(f"BLEU Score for m2m100: {bleu_m2m.score}")
print(f"BLEU Score for mbart: {bleu_mbart.score}")
```

Opus-MT:-  2.26

A BLEU score around 2 is relatively low. Opus-MT might be struggling with the translation task, possibly due to the complexity of the sentences, domain-specific language, or other issues.

m2m100:-  21.82

A BLEU score of 21.82 is quite high, suggesting good translation quality. m2m100 is performing well on your dataset.

mbart:-  9.60

A BLEU score of 9.60 is also relatively low, indicating poor translation quality. mbart seems not to be performing well.
